### Data imputation

```python
url = 'https://vincentarelbundock.github.io/Rdatasets/csv/datasets/airquality.csv'
df = pd.read_csv(url)

# create a new copy of df
df_new = df.copy()

## compute the mean of the non-missing 'Ozone' values
mean_ozone = df['Ozone'].mean()

# replace missing values in 'Ozone' column with calculated mean
df_new['Ozone'] = df['Ozone'].fillna(mean_ozone)
df_new.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.kdeplot(df['Ozone'], label='Original Ozone (with NaNs)', fill=True)
sns.kdeplot(df_new['Ozone'], label='Imputed Ozone (mean)', fill=True)
plt.title('Density Plot of Ozone Before and After Mean Imputation')
plt.xlabel('Ozone')
plt.ylabel('Density')
plt.legend()
plt.show()
```

### Z-Score Normalization

```python
# import StandardScaler class
from sklearn.preprocessing import StandardScaler

# create an instance of the class Standard Scaler
scaler = StandardScaler()

# data to be scaled
X = df[['MedInc', 'AveOccup']]

# call scaler's function "fit" to compute mean and std deviation for scaling
scaler.fit(X)

# scaler has attributes .mean_ and .var_
print("Scaler's mean used for scaling is:")
display(scaler.mean_)

print("Scaler's variance used for scaling is:")
display(scaler.var_)
```

```python
# we can use scaler's "transform" function
# to perform z-score normalization
X_scaled = scaler.transform(X)
display(X_scaled)
```

```python
# scaler also has a function "fit_transform"
# that combines the two steps into one
new_scaler = StandardScaler()
X_alt = new_scaler.fit_transform(X)
```

```python
import numpy as np

are_equal = np.array_equal(X_scaled, X_alt)
print(f"Are X_scaled and X_alt identical? {are_equal}")
```

```python
# as we can see, the new columns have mean = 0 and variance = 1
df[['MedInc', 'MedInc_zScore', 'AveOccup', 'AveOccup_zScore']].describe()


# we can give X_scaled as new columns in our original DataFrame
df[['MedInc_zScore', 'AveOccup_zScore']] = X_scaled

# check the first 5 rows
df[['MedInc', 'MedInc_zScore', 'AveOccup', 'AveOccup_zScore']].head()
```

```python
sns.jointplot(data=df, x='MedInc_zScale', y='AveOccup_zScale', kind='scatter', s=5)
plt.suptitle('Joint Plot of MedInc_zScore vs. AveOccup_zScore', y=1.02)
plt.show()
```

```python
# Create the jointplot
g = sns.jointplot(data=df, x='MedInc_zScore', y='AveOccup_zScore', kind='scatter', s=5)

# Set the x and y limits for the main scatter plot (ax_joint) to zoom in
# You can adjust these values based on the area you want to focus on
g.ax_joint.set_xlim(-2, 3)
g.ax_joint.set_ylim(-.3, .3)

plt.suptitle('Zoomed-in Joint Plot of MedInc_zScore vs. AveOccup_zScore', y=1.02)
plt.show()
```

### Min-Max Scaling

```python
# import MinMaxScaler class
from sklearn.preprocessing import MinMaxScaler

# create an instance of Min-Max Scaling
scaler = MinMaxScaler()

# data to be scaled
X = df[['MedInc', 'AveOccup']]

# call scaler's "fit_transform function"
X_scaled = scaler.fit_transform(X)

# scaler has attributes .data_min_ and .data_max_
print("Scaler's min used for scaling")
print(scaler.data_min_)
print("Scaler's max used for scaling")
print(scaler.data_max_)

display(X_scaled)
```

```python
df[['MedInc_MinMaxScale', 'AveOccup_MinMaxScale']] = X_scaled

df[['MedInc', 'MedInc_MinMaxScale', 'AveOccup', 'AveOccup_MinMaxScale']].describe()
```

```python
sns.jointplot(data=df, x='MedInc_MinMaxScale', y='AveOccup_MinMaxScale', kind='scatter', s=5)
plt.suptitle('Joint Plot of MedInc_MinMaxScale vs. AveOccup_MinMaxScale', y=1.02)
plt.show()
```

```python
# Create the jointplot
g = sns.jointplot(data=df, x='MedInc_MinMaxScale', y='AveOccup_MinMaxScale', kind='scatter', s=5)

# Set the x and y limits for the main scatter plot (ax_joint) to zoom in
# You can adjust these values based on the area you want to focus on
g.ax_joint.set_xlim(-0.01, 0.75)
g.ax_joint.set_ylim(-0.0001, 0.004)

plt.suptitle('Zoomed-in Joint Plot of MedInc_MinMaxScale vs. AveOccup_MinMaxScale', y=1.02)
plt.show()
```

## Outlier Detection

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/jolineuichanco/DataAnalytics/master/demos/california_housing.csv'
df = pd.read_csv(url)

df.head()
```

```python
# Visualize the distribution of 'AveOccup' after dropping outliers
sns.histplot(df['AveOccup'], kde=True)
plt.title('Distribution of AveOccup')
plt.xlabel('Average Occupancy')
plt.ylabel('Frequency')
plt.show()
```

### Using IQR Method

```python
# calculate quantiles
Q1 = df['AveOccup'].quantile(0.25)
Q3 = df['AveOccup'].quantile(0.75)
IQR = Q3 - Q1

# define whiskers
lower_boundary = Q1 - 1.5*IQR
upper_boundary = Q3 + 1.5*IQR

# identify outliers
outliers = df[(df['AveOccup'] < lower_boundary) | (df['AveOccup'] > upper_boundary)]
display(outliers)
```

```python
# Drop the outliers from the original DataFrame
df_no_outliers = df.drop(outliers.index)

# Visualize the distribution of 'AveOccup' after dropping outliers
sns.histplot(df_no_outliers['AveOccup'], kde=True)
plt.title('Distribution of AveOccup After Outlier Removal (IQR Method)')
plt.xlabel('Average Occupancy')
plt.ylabel('Frequency')
plt.show()
```

```python
from scipy.stats import zscore

# Calculate Z-scores for each data point
z_scores = zscore(df['AveOccup'])

# Define a threshold for identifying outliers
threshold = 2

# Identify outliers
outliers = df[(z_scores > threshold) | (z_scores < -threshold)]
display(outliers)
```

```python
# Drop the outliers from the original DataFrame
df_no_outliers = df.drop(outliers.index)

# Visualize the distribution of 'AveOccup' after dropping outliers
sns.histplot(df_no_outliers['AveOccup'], kde=True)
plt.title('Distribution of AveOccup After Outlier Removal (z-Score Method)')
plt.xlabel('Average Occupancy')
plt.ylabel('Frequency')
plt.show()
```
